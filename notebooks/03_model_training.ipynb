```
import numpy as np
from pathlib import Path
from src.baseline_cnn import build_baseline_cnn, convert_patient_to_tensor
from src.load_subset_from_h5 import load_stratified_subset

hdf5_path = Path('../data/processed/signals_100hz_DEV.h5')
X_raw, y_train, _ = load_stratified_subset(hdf5_path, split='train', sample_fraction=0.05)
X_raw_val, y_val, _ = load_stratified_subset(hdf5_path, split='val', sample_fraction=0.05)

print(f"Konwersja {len(X_raw)} próbek treningowych do CWT")
X_cwt = np.array([convert_patient_to_tensor(sig) for sig in X_raw])
X_cwt_val = np.array([convert_patient_to_tensor(sig) for sig in X_raw_val])

model = build_baseline_cnn(input_shape=(39, 1000, 12), num_classes=5)
history = model.fit(
    X_cwt, y_train,
    validation_data=(X_cwt_val, y_val),
    epochs=3,
    batch_size=16,
    verbose=1
)
```
